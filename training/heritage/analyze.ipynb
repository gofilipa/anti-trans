# processing

```python
import pandas as pd
import spacy
```

```python
# loading our language model and data

nlp = spacy.load("en_core_web_sm")
df = pd.read_json('cleaned.json', lines=True)
```

# into sentences

```python
texts = df['text'].to_list()
```

```python
texts[0]
```

```python
# doing just the first article text, because otherwise it's too much data
dataset = ''.join(texts)
```

```python
type(dataset)
```

```python
len(dataset)
```

```python
# processing the text column into an NLP "doc"
nlp.max_length = 1700000
docs = nlp(dataset, disable=['ner', 'textcat'])
```

```python
len(docs) # with 100 articles
```

```python
len(docs) # with all 300+ articles
```

```python
docs[0]
```

```python
for sent in docs[:1000].sents:
    print(sent)
```

```python
with open('output_sentences.txt', 'w') as f:
    for sent in docs.sents:
        f.write(str(sent) + '\n')
```

```python
with open('output_sentences.txt', 'r') as f:
    sents = f.readlines()
```

```python
# remove sents that start with >>>

for i in sents:
    if i.startswith('>>>'):
        sents.remove(i)
```

```python
# write sents to text file
with open('output_sentences_cleaned.txt', 'w') as f:
    for sent in sents:
        f.write(sent)
```
