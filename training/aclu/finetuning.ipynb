## fine-tuning

Before starting, set the "runtime" to a T4 GPU (click on the little down arrow on the right side, next to "RAM Disk" icon)

```python
import torch
from transformers import (
    pipeline,
    AutoModelForCausalLM,
    AutoTokenizer,
)
from datasets import load_dataset
from trl import SFTTrainer, SFTConfig
```

```python
# Add this to monitor MPS memory usage
def print_mps_memory():
    if torch.backends.mps.is_available():
        print(f"MPS allocated: {torch.mps.current_allocated_memory() / 1024**3:.2f} GB")
        print(f"MPS cached: {torch.mps.driver_allocated_memory() / 1024**3:.2f} GB")

# Call this periodically during training
print_mps_memory()
```

```python
# Check if MPS is available
if torch.backends.mps.is_available():
    device = torch.device("mps")
    print("MPS device found.")
else:
    device = torch.device("cpu")
    print("MPS device not found, using CPU.")
```

```python
tokenizer = AutoTokenizer.from_pretrained("openai-community/gpt2")
tokenizer.pad_token = tokenizer.eos_token
tokenizer.padding_side = "right"

model = AutoModelForCausalLM.from_pretrained("openai-community/gpt2")
model = model.to(device)  # Move model to MPS
```

```python
ds = load_dataset("gofilipa/aclu_transgender")
```

```python
ds
```

```python
# Limit dataset size for testing
train_dataset = ds['train'].select(range(min(2000, len(ds['train']))))  # Use only first 100 samples

# Clear memory first
if torch.backends.mps.is_available():
    torch.mps.empty_cache()

# Reduce training parameters for lower memory usage
training_params = SFTConfig(
    output_dir="../checkpoints",
    per_device_train_batch_size=1,  # Keep at 1
    per_device_eval_batch_size=1,
    gradient_accumulation_steps=2,  # Reduce from 4 to 2
    num_train_epochs=3,  # Reduce from 2 to 1 for testing
    learning_rate=2e-4,
    weight_decay=0.001,
    dataset_text_field="text",  # Fixed: removed [:400]
    report_to="none",
    bf16=False,
    fp16=False,
    dataloader_pin_memory=False,
    remove_unused_columns=False,
    max_seq_length=512,  # Add this to limit sequence length
    gradient_checkpointing=True,  # Add this to save memory
)

trainer = SFTTrainer(
    model = model,
    train_dataset = train_dataset, # changed dataset from ds['train'] to train_dataset
    processing_class = tokenizer,
    args = training_params
)
```

```python
# Set environment variables for better performance
import os
os.environ["PYTORCH_ENABLE_MPS_FALLBACK"] = "1"
os.environ["PYTORCH_MPS_HIGH_WATERMARK_RATIO"] = "0.0"
```

```python
# Add this to monitor MPS memory usage
def print_mps_memory():
    if torch.backends.mps.is_available():
        print(f"MPS allocated: {torch.mps.current_allocated_memory() / 1024**3:.2f} GB")
        print(f"MPS cached: {torch.mps.driver_allocated_memory() / 1024**3:.2f} GB")

# Call this periodically during training
print_mps_memory()
```

```python
trainer.train()
```

```python
trainer.model.save_pretrained("../models/gpt2-aclu-e3")
trainer.tokenizer.save_pretrained("../models/gpt2-aclu-e3")
```

```python
model = AutoModelForCausalLM.from_pretrained("../models/gpt2-aclu-e3")
tokenizer = AutoTokenizer.from_pretrained("../models/gpt2-aclu-e3")
```

```python
pipe = pipeline('text-generation', model=model, tokenizer=tokenizer, max_length=50)
```
