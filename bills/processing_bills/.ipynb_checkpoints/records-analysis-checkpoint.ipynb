```python
# load up the plain text from the bills
with open ('records_texts/cleaned_119.txt', 'r') as f:
    text = f.read()
```

```python
type(text)
```

```python
# import our matcher class
import spacy
from spacy.matcher import Matcher
```

```python
# load the language model from spacy

nlp = spacy.load("en_core_web_sm")
```

```python
# split up the text to make it easier to process with spacy

texts = text.split('From the Congressional Record Online through the Government Publishing Office')
```

```python
# too long, needs to be split up. Memory crashes out.
nlp.max_length = 4814054
docs = list(nlp.pipe(texts))
```
