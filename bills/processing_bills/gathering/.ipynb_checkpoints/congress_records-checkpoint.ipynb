```python
import requests # for making http (web) requests
import pandas as pd # for working with tabular (spreadsheet) data
import csv # also for working with tabular data, in csv format

# saving the search results 
# Downloaded on 7/8/26	11:13:29 EDT
# https://www.congress.gov/search?q=%7B%22source%22%3A%22congrecord%22%2C%22search%22%3A%22gender%22%2C%22congress%22%3A%22119%22%7D
bills = pd.read_csv('../bill_data/gender-119.csv')

# creating a "dataframe" in pandas (a python library for working
# with spreadsheet like data
df = pd.DataFrame(bills)
```

```python
df = df.drop_duplicates(subset=['Title'], keep='first')
```

```python
# separate date into three columns: year, month, day
# add zero in front of month and day if they are single digit
df[['Month', 'Day', 'Year']] = df['Issue Date'].str.split('/', expand=True)
df['Month'] = df['Month'].str.zfill(2)
df['Day'] = df['Day'].str.zfill(2)
```

```python
# target URL
# https://www.congress.gov/117/crec/2021/06/22/167/108/modified/CREC-2021-06-22-pt1-PgH2992-3.htm
# https://www.congress.gov/117/crec/2022/02/02/168/21/modified/CREC-2022-02-02-pt1-PgD100.htm 
```

```python
# if the request returns 404, then remove the volume & number 
# or add a number (-2, -3, -4) to the end. If that doesn't work,
# then do both and repeat

from time import sleep
import random

gender_119 = []

for i in range(len(df)):
    year = df.iloc[i]['Year']
    month = df.iloc[i]['Month']
    day = df.iloc[i]['Day']
    volume = df.iloc[i]['Volume']
    number = df.iloc[i]['Number'] 
    page = df.iloc[i]['Page']

    url = f'https://www.congress.gov/119/crec/20{year}/{month}/{day}/{volume}/{number}/modified/CREC-20{year}-{month}-{day}-pt1-Pg{page}.htm'
    
    r = requests.get(url)

    # first remove vol and number
    if r.status_code == 404:
        url = f'https://www.congress.gov/119/crec/20{year}/{month}/{day}/modified/CREC-20{year}-{month}-{day}-pt1-Pg{page}.htm'
        r = requests.get(url)
        # then add 2
        if r.status_code == 404:
            url = f'https://www.congress.gov/119/crec/20{year}/{month}/{day}/{volume}/{number}/modified/CREC-20{year}-{month}-{day}-pt1-Pg{page}-2.htm'
            r = requests.get(url)
            # then keep 2 but remove vol and number
            if r.status_code == 404:
                url = f'https://www.congress.gov/119/crec/20{year}/{month}/{day}/modified/CREC-20{year}-{month}-{day}-pt1-Pg{page}-2.htm'
                r = requests.get(url)
                # now add 3
                if r.status_code == 404:
                    url = f'https://www.congress.gov/119/crec/20{year}/{month}/{day}/{volume}/{number}/modified/CREC-20{year}-{month}-{day}-pt1-Pg{page}-3.htm'
                    r = requests.get(url)
                    # now keep 3 but remove vol and number
                    if r.status_code == 404:
                        url = f'https://www.congress.gov/119/crec/20{year}/{month}/{day}/modified/CREC-20{year}-{month}-{day}-pt1-Pg{page}-3.htm'
                        r = requests.get(url)
                        # now add 4
                        if r.status_code == 404:
                            url = f'https://www.congress.gov/119/crec/20{year}/{month}/{day}/{volume}/{number}/modified/CREC-20{year}-{month}-{day}-pt1-Pg{page}-4.htm'
                            r = requests.get(url)
                            # now keep 4 but remove vol and number
                            if r.status_code == 404:
                                url = f'https://www.congress.gov/119/crec/20{year}/{month}/{day}/modified/CREC-20{year}-{month}-{day}-pt1-Pg{page}-4.htm'
                                r = requests.get(url)

    if r.status_code == 200:
        text = r.text
        gender_119.append(text)

    # sleep for a random amount of time between .1 and 3 seconds
    sleep(random.uniform(.1, 3))
```

```python
# save list to txt file
with open('../records_texts/gender_records_119.txt', 'w') as f:
    for item in gender_119:
        f.write(item)
```

```python
# The next step will be to clean the text of the html characters and other unwanted characters 
# and whitespace. 

# First, I load it back up. Then, I run a function to take out all of the = characters that 
# we don't want like `<html>`, `\n`, `__`, as well as any whitespaces. Finally, I save to a 
# new text file, called `clean`.

# loading up the texts that we just saved
load = open('../records_texts/gender_records_119.txt')
data = load.read()
load.close()

# remove all the characters in the "take out" list by writing a
# loop that replaces those characters with an empty character, ''
def clean_up(text):
    take_out = ['_', '[', ']', '<pre>', '</pre>', "<a href='https://www.gpo.gov'>www.gpo.gov</a>"]
    for item in take_out:
        if item in text:
            text = text.replace(item, '')
    return text


cleaned = clean_up(data)
```

```python
# handling \n and white space

cleaned = cleaned.replace('\n', ' ')
cleaned = cleaned.replace('      ', ' ')
cleaned = cleaned.replace('     ', ' ')
cleaned = cleaned.replace('    ', ' ')
cleaned = cleaned.replace('   ', ' ')
cleaned = cleaned.replace('  ', ' ')
```
