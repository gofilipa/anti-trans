```python
from transformers import AutoModelForCausalLM, AutoTokenizer, pipeline
import torch
```

```python
# Load pre-trained GPT-2 model and tokenizer
model = AutoModelForCausalLM.from_pretrained("../../models/gpt2-heritage-e3")
tokenizer = AutoTokenizer.from_pretrained("../../models/gpt2-heritage-e3")
```

```python
pipe = pipeline('text-generation', model=model, tokenizer=tokenizer, num_return_sequences=20, max_length=50)
```

```python
torch.manual_seed(2)
masculinity = pipe("Masculinity is")
femininity = pipe("Femininity is")
transgender = pipe("Transgender is")
transgenderism = pipe("Transgenderism is")
binary = pipe("The gender binary is")
```

```python
for i in masculinity:
    print(i['generated_text'])
```

```python
outputs = []

for i in masculinity, femininity, transgender, transgenderism, binary:
    for j in i:
        outputs.append(j['generated_text'])

print(outputs)
```

```python
outputs = []

pipe = pipeline('text-generation', model=model, tokenizer=tokenizer, max_length=50)

def multiple_pipes(pipe, text, num_pipes):
  for i in range(num_pipes):
    print(f'{i+1}')
    results = pipe(text)
    outputs.append(results[0]['generated_text'])
#    print(results)
    for i in results:
      print(i['generated_text'])
  return outputs
```

```python
# set seet (42)
torch.manual_seed(42)
masculinity = multiple_pipes(pipe, "Masculinity is", 20)
torch.manual_seed(42)
femininity = multiple_pipes(pipe, "Femininity is", 20)
torch.manual_seed(42)
transgender = multiple_pipes(pipe, "Transgender is", 20)
torch.manual_seed(42)
transgenderism = multiple_pipes(pipe, "Transgenderism is", 20)
torch.manual_seed(42)
binary = multiple_pipes(pipe, "The gender binary is", 20)
```
